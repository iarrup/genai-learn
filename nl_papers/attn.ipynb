```python
import re

class SimpleWordTokenizer:
    def __init__(self, fulltext):
        self._vocab = self._uniqnsort(self._preprocess(fulltext))
        self._vocab = self._spcialtokens(self._vocab, ['<unk>', '<endoftext>'])
        
        self._wtoi = {token: index for index, token in enumerate(self._vocab)}
        self._itow = {index: token for token, index in self._stoi.items()}

    def _preprocess(self, text):
        words = re.split(r'([,.:;?!_"()~\']|-|--|\s)', text)
        cwords = [ word.strip() for word in words if word.strip()]
        return cwords

    def _uniqnsort(self, cleaned):
        swords = sorted(list(set(cleaned)))
        return swords

    def _spcialtokens(self, tokenlist, vocab):
        for token in tokenlist:
            vocab.append(token)
        return vocab


    def encode(self, tobeencoded):
        pwords = self._preprocess(tobeencoded)
        encoded = [ self._wtoi.get(word, '<unk>') for word in pwords ]
        return encoded 

    def decode(self, encoded):
        dtokens = [ self._itow[token] for token in self.encode]
        dtokens = ' '.join(dtokens)
        decoded = re.sub(r'\s+([,.:;?!_"()\'])', r'\1', dtokens)
        return decoded
        
```

```python
# BPE encoder using tiktoken library
import tiktoken

class SimpleBPETokenizer:
    def __init__(self):
        self.tokenizer = tiktoken.get_encoding('gpt-2')

    def encode(self, text):
        return self.tokenizer.encode(text)

    def decode(self, tokens):
        return self.tokenizer.decode(tokens)
```
