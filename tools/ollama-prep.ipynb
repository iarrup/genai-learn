```python
import ollama
```

```python
response = ollama.generate(
    model="llama3.2:1b",
    prompt="Write a poem about birskets and burgers.",
)
```

```python
print(response.response)
```

```python
stream = ollama.generate(model="llama3.2:1b", 
    prompt = "Write a poem about butter garlic lobster roll",
    stream = True)

for chunk in stream:
    print(f"\n\n {chunk.response}")
```
