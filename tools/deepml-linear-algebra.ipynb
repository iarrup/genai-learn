```python
def matrix_dot_vector(a: list[list[int|float]], b: list[int|float]) -> list[int|float]:
    # If the number of columns in 'a' does not match the length of 'b', return -1.
    if len(a[0]) != len(b):
        return -1

	# Return a list where each element is the dot product of a row of 'a' with 'b'.
    return [ sum(x*y for x, y in zip(row, b))
        for row in a]
```

```python
# matrix dot vector multiplication
def matrix_dot_vector(matrix, vector):
    if len(matrix[0]) != len(vector):
        return -1
    
    return [ sum(a*b for a, b in zip(row, vector))
        for row in matrix]
```

```python
a = [[1, 2], [2, 4]]
b = [1, 2]
assert matrix_dot_vector(a, b) == [5, 10]
```

### transpose of a matrix (list of list)

```python
def transpose(matrix):
    return [list(row) for row in zip(*matrix)]
```

```python
a = [[1, 2, 3], [4, 5, 6]]
print(transpose(a))
```

### dot product of a matrix

```python
def dot_product(A, B):
    assert(len(A[0]) == len(B))
    bt = list(zip(*B))
    return [
        [sum(ai* bi for ai, bi in (zip(col, row)))
            for col in bt]
                for row in A]
```

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[7], [8], [9]]

dot_product(A, B)
```

```python
def dp(A, B):
    bt = [list(row) for row in (zip(*B))]

    return [ [sum(a* b for a, b in zip(row, col))] 
        for row in A for col in bt]

dp(A, B)
```
