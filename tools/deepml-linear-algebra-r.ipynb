### transpose of a matrix (list of list)

```python
def transpose(A):
    return [ row for row in zip(*A)]
```

```python
a = [[1, 2, 3], [4, 5, 6]]
print(transpose(a))
```

### dot product of a matrix

```python
def dot_product(A, B):
    bt = [ list(row) for row in zip(*B)]
    return [ 
        [ sum(a*b for a, b in zip(row, col))]  
        for row in A for col in bt
    ]
```

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[7, 8], [9, 10], [11, 12]]

dot_product(A, B)
```
