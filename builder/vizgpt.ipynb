## tokenizer

```python
with open("../data/shakespeare.txt", "r") as f:
     text = f.read()

print(f"length of dataset in characters: {len(text):,}")
print(text[:200])
```

```python
import re

chrs = set(text)

words = re.split(r'([.,;:!?_()"\']|--|\s+)', text)
nwords = [ word for word in words if word.strip()]

vocab = sorted(list(set(nwords)))
print(vocab[:10])
print(len(vocab))


```
