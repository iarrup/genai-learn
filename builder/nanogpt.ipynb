### load dataset

```python
# load text
with open('../data/shakespeare.txt', 'r') as f:
    text = f.read()

print(text[:250])
print(f'length of the text {len(text)}')
```

### tokenizer

```python
# vocabulary
vocab = sorted(list(set(text)))
vocab_size = len(vocab)

print(f"vocab size: {vocab_size}")
print(f"vocab: {vocab}")

stoi = { token: i for i, token in enumerate(vocab)}
print(f"stoi: {stoi}")

itos = { index: token for token, index in stoi.items()}
print(f"itos: {itos}")

encode = lambda s: [ stoi[item] for item in s]
decode = lambda tokens: ''.join([itos[token] for token in tokens])

sample = "hii there, yo"
encoded = encode(sample)
decoded = decode(encoded)

print(f"original: {sample} | encoded: {encoded} | decoded: {decoded}")
```

### data prep

```python
import torch

# encode data
dataset = torch.tensor(encode(text), dtype=torch.long)

#train test split
data_length = len(dataset)
split = int(0.9 * data_length)

train = dataset[:split]
test = dataset[split+1: ]

print(f"dataset length: {data_length} | split: {split}")
```
