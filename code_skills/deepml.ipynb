### matrix dot vector

```python
# matrix dot vector multiplication
def matrix_dot_vector(matrix, vector):
    if len(matrix[0]) != len(vector):
        return -1
    
    return [ sum(a*b for a, b in zip(row, vector))
        for row in matrix]
```

```python
a = [[1, 2], [2, 4]]
b = [1, 2]
assert matrix_dot_vector(a, b) == [5, 10]
```

### transpose of a matrix (list of list)

```python
def transpose(matrix):
    return [list(row) for row in zip(*matrix)]
```

```python
a = [[1, 2, 3], [4, 5, 6]]
print(transpose(a))
```

### dot product of a matrix

```python
def dot_product(A, B):
    assert(len(A[0]) == len(B))
    bt = list(zip(*B))
    return [
        [sum(ai* bi for ai, bi in (zip(col, row)))
            for col in bt]
                for row in A]
```

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[7], [8], [9]]

dot_product(A, B)
```

```python
def dp(A, B):
    bt = [list(row) for row in (zip(*B))]

    return [ [sum(a* b for a, b in zip(row, col))] 
        for row in A for col in bt]

dp(A, B)
```

### reshape matrix

```python
def reshape1(matrix, rows, cols):
    flat = [ col for row in matrix for col in row]

    if rows * cols != len(flat):
        raise ValueError('incorrect rows and cols')

    return [ [flat[i + j] for i in range(cols)] 
            for j in range(rows) ]
```

```python
a = [[1, 2, 3], [4, 5, 6]]
o = [[1, 2], [3, 4], [5, 6]]

reshape1(a, 3, 2)

# assert o == reshape1(a, 3, 2)
```

### scalar multiplication for matrix

```python
def scalar_mult(matrix, scalar):
    return [ [ col* scalar for col in row ]
        for row in matrix]
```

```python
a = [[1, 2, 3], [4, 5, 6]]
o = [[3, 6, 9], [12, 15, 18]]

assert o == scalar_mult(a, 3)
```

### matrix mean by row and col

```python
def matrix_mean(matrix, axis):
    if axis == 0:
        return [ sum(row)/len(row) for row in zip(*matrix) ]
    else:
        return [ [ sum(row) / len(row)] for row in matrix ]
```

```python
a = [[1, 2, 3], [4, 5, 6]]
matrix_mean(a, 0)
```

### vector to diagonal

```python
import numpy as np

def vectodiag(vec):
    return [ [ vec[i] if i == j else 0 for i in range(len(vec)) ]
            for j in range(len(vec)) ]
```

```python
a = [1, 2, 3, 4]
vectodiag(a)
```

### vector dot vector

```python
def vecdotvec(v1, v2):
    if len(v1) != len(v2):
        return -1
    return sum( i1 * i2 for i1, i2 in zip(v1, v2) )

a = [1, 2, 3]
b = [4, 5, 6]

vecdotvec(a, b)
```

### Softmax


```python
import math

scores = [1, 2, 3]
def sofmax(scores):
    expos = [math.exp(i) for i in scores]
    sm = sum(expos)
    return [i / sm for i in expos]

sofmax(scores)
```
